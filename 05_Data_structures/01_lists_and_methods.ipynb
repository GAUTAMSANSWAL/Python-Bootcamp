Lists are ordered, mutable (changeable) collections of items.

```python
# Creating a list

numbers = [1, 2, 3, 4, 5]
mixed = [10, "hello", 3.14]

print(numbers)
print(mixed[1])
```

```python
# Common list methods

my_list = [1, 2, 3]
second_list = [4, 5, 6]

my_list.append(4)   # [1, 2, 3, 4]
my_list.insert(1, 99)  # [1, 99, 2, 3, 4]
my_list.remove(2)   # [1, 99, 3, 4]
my_list.pop()       # Removes last element -> [1, 99, 3]
my_list.reverse()   # [3, 99, 1]
my_list.sort()      # [1, 3, 99]

print(my_list)

my_list.extend(second_list)  # [1, 3, 99, 4, 5, 6]
print(my_list)
```

```python
# List comprehensions

squared = [x**2 for x in range(5)]
print(squared)  # Output: [0, 1, 4, 9, 16]
```
