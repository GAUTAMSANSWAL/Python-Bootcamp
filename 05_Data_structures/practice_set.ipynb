1. Introduction to Lists

a) Create a list fruits = ["apple", "banana", "cherry"].

Print the first fruit.
Replace "banana" with "orange".
Print the length of the list.

b) Create a list of numbers from 1 to 10.

Print the first three numbers using slicing.
Print the last three numbers using slicing.

2. List Methods

a) Start with numbers = [5, 2, 9, 1, 7] and do the following:

Sort the list in ascending order.
Append the number 10 to the list.
Remove the number 2 from the list.

b) Create a list names = ["Alice", "Bob", "Charlie"] and use the insert() method to add "David" at index 1.

```python
numbers = [5, 2, 9, 1, 7]

numbers.sort()
print(numbers)
numbers.append(10)
print(numbers)
numbers.remove(2)
print(numbers)

names = ["Alice", "Bob", "Charlie"]

print(names)
names.insert(1, "David")
print(names)
```

b) Try to modify the tuple by setting coordinates[0] = 50 — note what happens.

c) Convert the tuple to a list, change its first element to 50, and convert it back to a tuple.

4. Sets and Set Methods

a) Create a set my_set = {1, 2, 3, 3, 4} and print it. (What happens to duplicate 3?)

b) Add 5 to the set, remove 2, and check if 4 is in the set.

c) Create two sets:

a = {1, 2, 3}
b = {3, 4, 5}

Find their:

Union

Intersection

Difference (a - b)

```python
my_set = {1, 2, 3, 3, 4}
print(my_set)
# duplicate 3 gets ignored and only one unique value is stored in the set

my_set.add(5)
print(my_set)
my_set.remove(2)
print(my_set)
if 4 in my_set:
    print("4 is in the set")

a = {1, 2, 3}
b = {3, 4, 5}

# union
print(a.union(b))
# intersection
print(a.intersection(b))
# difference
print(a.difference(b))
```

5. Dictionaries and Dictionary Methods
a) Create a dictionary student = {"name": "John", "age": 20, "grade": "A"} and:

Print the value of "name".

Change "grade" to "A+".

Add a new key "city" with value "Delhi".

b) Create a dictionary of three friends and their phone numbers. Use:

keys() to get all names

values() to get all numbers

items() to loop over key-value pairs and print them

```python
student = {"name": "John", "age": 20, "grade": "A"}

print(student["name"])
student["grade"] = "A+"
print(student)
student["city"] = "Delhi"
print(student)

dict_names_numbers = {"Alice": 123, "Bob": 456, "Charlie": 789}

print(dict_names_numbers.keys())
print(dict_names_numbers.values())
print(dict_names_numbers.items())
```
